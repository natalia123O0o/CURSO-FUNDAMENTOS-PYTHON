 2.
 En un partido de fútbol, se ofrece un descuento a los aficionados que depende del estrato y la edad. Si el estrato es 1 y su edad es menor a 18 el descuento será del 20% sobre el valor de la boleta. Si el estrato es 1 y el alumno tiene 18 o mas años, el descuento será del 15%. Si el estrato es 2 y la edad es menor a 18 años, el descuento será del 10% y si el estrato es 2 y la edad es 18 años o más, el descuento será del 5%. Determinar el total del dinero recaudado y descontado por las últimas N personas que ingresan al partido.

```python
# Solicitar al usuario que ingrese el precio de la boleta (convertido a entero)
precio_boleta = int(input("Ingrese el valor de la boleta: "))
# Solicitar al usuario que indique el número de personas (convertido a entero)
num_personas = int(input("Ingrese el número de personas: "))

# Inicializar variables:
# 'total_recaudado' almacena el monto total recibido por las boletas vendidas.
# 'total_descuento' almacena la suma total de los descuentos aplicados.
total_recaudado = 0
total_descuento = 0

# Se utiliza un bucle 'for' para procesar cada persona, desde 0 hasta el número de personas (exclusivo).
for i in range(num_personas):
    print(f"\nPersona {i + 1}:")  # Mostrar al usuario qué persona se está procesando.

    # Solicitar al usuario que ingrese el estrato socioeconómico (debe ser 1 o 2).
    estrato = int(input("Ingrese el estrato (1 o 2): "))
    # Solicitar al usuario que ingrese la edad.
    edad = int(input("Ingrese la edad: "))

    # Evaluar las condiciones para calcular el descuento:
    if estrato == 1:  # Si el estrato es 1
        descuento = precio_boleta * (0.20 if edad < 18 else 0.0)  # 20% si menor de 18, sino 0%.
    elif estrato == 2:  # Si el estrato es 2
        descuento = precio_boleta * (0.15 if edad < 18 else 0.10)  # 15% si menor de 18, 10% si mayor.
    else:
        # Si el estrato ingresado no es válido, asignar descuento como 0 y notificar.
        print("Estrato no válido. No se aplicará descuento.")
        descuento = 0.0

    # Calcular el precio final restando el descuento al precio de la boleta.
    precio_final = precio_boleta - descuento

   # Acumular el precio final en el total recaudado
# A 'total_recaudado' se le suma el valor de 'precio_final',
# que representa el monto total recaudado tras procesar a cada persona.
total_recaudado += precio_final

# Acumular el descuento aplicado en el total de descuentos
# A 'total_descuento' se le suma el valor de 'descuento',
# lo cual permite llevar un registro de todos los descuentos aplicados hasta el momento.
total_descuento += descuento

# Mostrar al usuario el descuento aplicado en esta iteración (para esta persona).
# Se usa 'print()' para desplegar el mensaje de "Descuento aplicado" junto al monto formateado como número entero.
print(f"Descuento aplicado: ${descuento:.0f}")

# Mostrar al usuario el precio final que debe pagar esta persona después de aplicar el descuento.
# Nuevamente, se utiliza 'print()' para mostrar el mensaje "Precio final" con el valor formateado como número entero.
print(f"Precio final: ${precio_final:.0f}")


# Al finalizar el bucle, mostrar un resumen de los totales acumulados:
print("\nResumen Final:")
print(f"Total recaudado: ${total_recaudado:.0f}")  # Imprimir el total recaudado (precio final total).
print(f"Total descontado: ${total_descuento:.0f}")  # Imprimir el total de descuentos aplicados.
```

3.  Leer un password de ingreso a un programa y mostrar el mensaje de bienvenida si es correcto. Mientras no lo sea, debe mostrar el mensaje de Password incorrecto. El programa debe terminar automáticamente al quinto intento fallido.

```python
# Definir la contraseña correcta
# 'password_corecto' almacena el password que el usuario debe ingresar para acceder.
password_corecto = 'Gato123'

# Inicializar el contador de intentos fallidos
# 'intentos_fallidos' llevará la cuenta de cuántas veces se falla al ingresar el password.
intentos_fallidos = 0

# Mientras los intentos fallidos sean menores a 5, el usuario podrá seguir intentando ingresar el password.
while intentos_fallidos < 5:
    # Solicitar al usuario que ingrese el password
    # Se captura el password ingresado a través de la función 'input()'.
    password = input('ingrese el password')

    # Verificar si el password ingresado coincide con el correcto
    if password == password_corecto:
        # Si el password es correcto, se muestra un mensaje de éxito
        print('es correcto su password')
        # Finalizar el programa saliendo del bucle con 'break'.
        break
    else:
        # Si el password es incorrecto, se muestra un mensaje de error
        print('es incorrecto su password')
        # Incrementar el contador de intentos fallidos en 1
        intentos_fallidos += 1

# Si el usuario ha fallado 5 veces, el bucle termina y se muestra un mensaje final
if intentos_fallidos == 5:
    # Indicar que el límite de intentos ha sido superado
    print('ha superado el limite de los 5 intentos')
```
